```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans
import seaborn as sns
```

```python
np.random.seed(42)
numeros = np.random.rand(100,2)
numeros
```

```python
modelo = KMeans(n_clusters=5, random_state=42)
modelo.fit(numeros)
```

```python
etiquetas = modelo.labels_
centroides = modelo.cluster_centers_
centroides
```

```python
etiquetas
```

```python
plt.figure(figsize=(10,6))
plt.scatter(numeros[:,0],numeros[:,1], marker='o', c=etiquetas)
plt.scatter(centroides[:,0],centroides[:,1], marker='x', s=200, color='red', label='centroides')
plt.legend()
plt.show()
```

```python
import numpy as np 
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

np.random.seed(42)

datos = np.random.rand(100,2)

kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(datos)

etiquetas = kmeans.labels_
centroides = kmeans.cluster_centers_

plt.figure(figsize=(10,6))
plt.scatter(datos[:,0], datos[:,1], c=etiquetas,  marker='o', cmap='viridis')
plt.scatter(centroides[:,0], centroides[:,1], c='red', s=200, marker='x', label='Centroides')
plt.title('K-Means Clustering')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
import seaborn as sns
```

```python
pinguinos = sns.load_dataset("penguins")
pinguinos.head()
```

```python
pinguinos.info()
```

```python
pinguinos['species'].unique()
```

```python
pinguinos = pinguinos.dropna(axis=0)
```

```python
pinguinos.info()
```

```python
data = pinguinos[["bill_length_mm","bill_depth_mm"]]
data
```

```python
model = KMeans(n_clusters=4, random_state=42)
model.fit(data)
```

```python
etiquetas = model.labels_
centroides = model.cluster_centers_
print(etiquetas)
print(centroides)
```

```python
plt.figure(figsize=(20,10))
plt.scatter(data.iloc[:,0], data.iloc[:,1], c=etiquetas, marker='o')
plt.scatter(centroides[:,0], centroides[:,1], marker='x', c='red', s=200)
```

```python
df = sns.load_dataset("iris")
df.head(5)
```

```python
df.info()
```

```python
X1 = df.values[:,0]
X2 = df.values[:,1]
data = df[['sepal_length', 'sepal_width']]
data
```

```python
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(data)
```

```python
labels = kmeans.labels_
centroids = kmeans.cluster_centers_
```

```python
plt.figure(figsize=(10,10))
plt.scatter(X1, X2, c=labels, marker='o')
plt.scatter(centroids[:,0], centroids[:,1], c='red', marker='x', s=200 )
```
