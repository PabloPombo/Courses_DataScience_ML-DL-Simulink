```python
######################### Unittest ##############################




class base_test(unittest.TestCase):
    
    def setUp(self):
        self.driver = webdriver.Chrome()
        self.t = 1
        
    
    def elements(self):   
        driver = self.driver
        driver.get("https://www.saucedemo.com/")  
        driver.maximize_window()
        time.sleep(self.t)
        
        name = driver.find_element(By.XPATH, "//input[@id='user-name']")
        password = driver.find_element(By.XPATH, "//input[@id='password']")
        btn = driver.find_element(By.XPATH, "//input[@id='login-button']")
        
        return name, password, btn
    
    
    def test_login1(self):
        # WRONG NAME AND PASSWORD
        driver = self.driver
        
        name, password, btn = self.elements()
        
        name.send_keys("Pablo")
        password.send_keys("password")
        btn.click()
        
        error = driver.find_element(By.XPATH, "//h3[@data-test='error']")
        error = error.text
        
        if error == "Epic sadface: Username and password do not match any user in this service":
            print("Username and password do not match")
            print("Frist error OK")
        
        time.sleep(self.t)
        
        
    def test_login2(self):
        # NO NAME
        driver = self.driver
        
        name, password, btn = self.elements()
        
        #name.send_keys("Pablo")
        password.send_keys("password")
        btn.click()
        
        error = driver.find_element(By.XPATH, "//h3[@data-test='error']")
        error = error.text
        
        if error == "Epic sadface: Username is required":
            print("Username not added")
            print("Second error OK")
        
        time.sleep(self.t)   


    def test_login3(self):
        # NO PASSWORD
        driver = self.driver

        name, password, btn = self.elements()

        name.send_keys("Pablo")
        #password.send_keys("password")
        btn.click()
        error = driver.find_element(By.XPATH, "//h3[@data-test='error']")
        error = error.text

        if error == "Epic sadface: Password is required":
            print("No password added")
            print("3th error OK")

        time.sleep(self.t)
            

    def test_login4(self):
        # NO PASSWORD, NO NAME
        driver = self.driver

        name, password, btn = self.elements()

        #name.send_keys("Pablo")
        #password.send_keys("password")
        btn.click()

        error = driver.find_element(By.XPATH, "//h3[@data-test='error']")
        error = error.text

        if error == "Epic sadface: Username is required":
            print("No password neither name added")
            print("4th error OK")

        time.sleep(self.t)
        

    def test_login5(self):
        # CORRECT WAY
        driver = self.driver

        name, password, btn = self.elements()

        name.send_keys("standard_user")
        password.send_keys("secret_sauce")
        btn.click()

        WebDriverWait(driver, 10).until(lambda d: "inventory.html" in d.current_url)
        assert "inventory.html" in driver.current_url
        print("✅ Login")          
        
    def tearDown(self):
        self.driver.close()
        

if __name__ == '__main__':
    #unittest.main()
    unittest.main(argv=['first-arg-is-ignored'], exit=False)
```
