```python
from selenium import webdriver 
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.service import Service
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.firefox.service import Service
from selenium.webdriver.firefox.options import Options
from selenium.webdriver.common.by import By

from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

from selenium.webdriver.support.ui import Select
from selenium.common.exceptions import TimeoutException


from webdriver_manager.chrome import ChromeDriverManager
import time
```

```python
######################### Backward and Forward ##############################
t = 2

driver = webdriver.Chrome()
driver.get("https://demoqa.com/text-box")
driver.maximize_window()
time.sleep(t)

driver.get("https://aulavirtual.aevirtualclass.com/login/index.php")
time.sleep(t)

driver.back()
time.sleep(t)

driver.get("https://github.com/PabloPombo/Courses_DataScience_ML-DL-Simulink/tree/main/Python-DS-ML_UDEMYCourse")
time.sleep(t)

driver.execute_script("window.history.go(-2)")
time.sleep(t)

driver.forward()
time.sleep(t)

driver.execute_script("window.history.go(1)")
time.sleep(t)

driver.close()
```

```python
######################### Implicit  wait ##############################
## Selenium to wait a certain amount of time when trying to find any element before throwing an error.
##It applies globally to all elements in your test.

t=2

driver = webdriver.Chrome()

driver.implicitly_wait(10) #will wait 10 s 
driver.get("https://demoqa.com/text-box")
driver.maximize_window()

driver.find_element(By.XPATH, "//input[@id='userName']").send_keys("Pablo Pombo")

time.sleep(t)

driver.close()
```

```python
######################### Explicit  wait ##############################
##You tell Selenium to wait for a specific condition to happen before moving on — for example, 
##“wait until this element is visible” or “clickable.”

from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC


t=2

driver = webdriver.Chrome()
driver.get("https://www.deutschlandstipendium.de/deutschlandstipendium/de/stipendium/haeufig-gestellte-fragen/haeufig-gestellte-fragen_node.html")
driver.maximize_window()

time.sleep(t)

btn = WebDriverWait(driver, 10).until(EC.element_to_be_clickable(
    (By.XPATH, "//*[@id='cookiebanner']/div/div/div/div[4]/div/p/button[1]")))

btn.click()

time.sleep(t)

driver.close()
```

```python
 ######################### List Box with TRY ##############################
from selenium.webdriver.support.ui import Select
from selenium.common.exceptions import TimeoutException

t=2
driver = webdriver.Chrome()
driver.get("https://demoqa.com/text-box")

driver.maximize_window()
time.sleep(t)


driver.find_element(By.XPATH, "//div[@class='element-list collapse show']//li[@id='item-3']").click()

driver.execute_script("window.scrollTo(0,800)")

try:
    rows = Select(WebDriverWait(driver,10).until(EC.element_to_be_clickable((By.XPATH, "//select[@aria-label='rows per page']"))))
    rows.select_by_index(0)
    time.sleep(t)
    rows.select_by_index(1)
    time.sleep(t)
    
#except TimeoutException as ex:  #Just time exception 
#    print(ex.msg)
#    print("El elemento no se encontro")
    
except Exception as ex:   #takes any kind of error 
    print("ERROR:", type(ex).__name__)
    print("Message:", ex)
    print("Element not found.")

driver.execute_script("window.scrollTo(0,-800)")

driver.find_element(By.XPATH, "//span[normalize-space()='Text Box']").click()

WebDriverWait(driver,10).until(EC.presence_of_element_located((By.XPATH, "//input[@id='userName']"))).send_keys("Pablo Pombo")
time.sleep(t)

driver.close()
```
