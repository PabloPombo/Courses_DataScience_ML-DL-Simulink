```python
from selenium import webdriver 
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.service import Service
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.firefox.service import Service
from selenium.webdriver.firefox.options import Options
from selenium.webdriver.common.by import By


from webdriver_manager.chrome import ChromeDriverManager
import time
```

```python
######################### With Xpath ##############################

driver = webdriver.Chrome()
driver.get("https://demoqa.com/text-box")

driver.maximize_window()
time.sleep(1)

driver.find_element(By.XPATH, "//input[@id='userName' and @type='text']").send_keys("Pablo Pombo")
time.sleep(1)
driver.find_element(By.XPATH, "//input[@id='userEmail']").send_keys("pablo-pombo@hotmail.com")
time.sleep(1)
driver.find_element(By.XPATH, "//textarea[@id='currentAddress']").send_keys("Dir uno")
time.sleep(1)
driver.find_element(By.XPATH, "//textarea[@id='permanentAddress']").send_keys("Dir dos")
time.sleep(1)
driver.execute_script("window.scrollTo(0, 500)")

driver.find_element(By.XPATH, "//button[@id='submit']").click()
time.sleep(1)

driver.quit()
```

```python
######################### With id ##############################

driver = webdriver.Chrome()
driver.get("https://demoqa.com/text-box")

driver.maximize_window()
time.sleep(1)

driver.find_element(By.ID, "userName").send_keys("Pablo Pombo")
time.sleep(1)
driver.find_element(By.ID, "userEmail").send_keys("pablo-pombo@hotmail.com")
time.sleep(1)
driver.find_element(By.ID, "currentAddress").send_keys("Dir uno")
time.sleep(1)
driver.find_element(By.ID, "permanentAddress").send_keys("Dir dos")
time.sleep(1)
driver.execute_script("window.scrollTo(0, 500)")

driver.find_element(By.ID, "submit").click()
time.sleep(1)

driver.quit()
```

```python
######################### With CSS ##############################

driver = webdriver.Chrome()
driver.get("https://demoqa.com/text-box")

driver.maximize_window()
time.sleep(1)

driver.find_element(By.CSS_SELECTOR, "#userName").send_keys("Pablo Pombo")
time.sleep(1)
driver.find_element(By.CSS_SELECTOR, "#userEmail").send_keys("pablo-pombo@hotmail.com")
time.sleep(1)
driver.find_element(By.CSS_SELECTOR, "#currentAddress").send_keys("Dir uno")
time.sleep(1)
driver.find_element(By.CSS_SELECTOR, "#permanentAddress").send_keys("Dir dos")
time.sleep(1)
driver.execute_script("window.scrollTo(0, 500)")

driver.find_element(By.CSS_SELECTOR, "#submit").click()
time.sleep(1)

driver.quit()
```

```python
######################### With Key.TAB ##############################

driver = webdriver.Chrome()
driver.get("https://demoqa.com/text-box")

driver.maximize_window()
time.sleep(1)

name = driver.find_element(By.CSS_SELECTOR, "#userName")
time.sleep(1)
name.send_keys("Pablo Pombo" + Keys.TAB + "pablo-pombo@hotmail.com" + Keys.TAB + "Dir uno" + Keys.TAB + "Dir dos")
time.sleep(1)

driver.execute_script("window.scrollTo(0, 500)")

driver.find_element(By.CSS_SELECTOR, "#submit").click()
time.sleep(2)

driver.quit()
```
