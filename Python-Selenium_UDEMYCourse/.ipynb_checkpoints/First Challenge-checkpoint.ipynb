```python
from selenium import webdriver 
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.service import Service
from selenium.webdriver.common.by import By

from selenium.webdriver.chrome.service import Service
from selenium.webdriver.firefox.service import Service
from selenium.webdriver.firefox.options import Options


from selenium.webdriver.support.ui import WebDriverWait, Select
from selenium.webdriver.support import expected_conditions as EC

from webdriver_manager.chrome import ChromeDriverManager
import time
```

```python
######################### First Challenge ##############################
t = 2
driver = webdriver.Chrome()
driver.get("https://demoqa.com/text-box")
driver.maximize_window()
time.sleep(t)

driver.find_element(By.XPATH, "//input[@id='userName']").send_keys("Pablo Pombo" + Keys.TAB + "pablo-pombo@hotmail.com")
driver.find_element(By.XPATH, "//textarea[@id='currentAddress']").send_keys("Dir 1" + Keys.TAB + "Dir 2")

try:
    btn1 = WebDriverWait(driver, 5).until(EC.element_to_be_clickable((By.XPATH, "//button[@id='submit']")))
    driver.execute_script("arguments[0].scrollIntoView()", btn1)
    btn1.click()
    
    print("Personal information been added")
    time.sleep(t)
    
except Exception as ex:
    print("ERROR:", type(ex).__name__)
    print("Message: ", ex)

driver.find_element(By.XPATH, "//span[normalize-space()='Upload and Download']").click()

try:
    image = WebDriverWait(driver,5).until(EC.element_to_be_clickable((By.XPATH, "//input[@id='uploadFile']")))
    image.send_keys("D:\Fotos-Imagenes\(Y).jpg")
    driver.find_element(By.XPATH, "//a[@id='downloadButton']").click()
    
    print("The image its been downloaded")
    time.sleep(t)
    
except Exception as ex: 
    print("ERROR:", type(ex).__name__)
    print("Message:", ex)

driver.find_element(By.XPATH, "//span[normalize-space()='Radio Button']").click()
driver.find_element(By.XPATH, "//label[@for='impressiveRadio']").click()
print("Liked the page")
time.sleep(t)


driver.find_element(By.XPATH, "//span[normalize-space()='Check Box']").click()

try:
    toggle = WebDriverWait(driver,5).until(EC.element_to_be_clickable((By.XPATH, "//button[@title='Toggle']//*[name()='svg']")))
    toggle.click()
    #driver.execute_script("arguments[0].click()",toggle)

    desktop = WebDriverWait(driver,5).until(EC.element_to_be_clickable((By.XPATH, "//li[@class='rct-node rct-node-parent rct-node-expanded']//li[1]//span[1]//button[1]//*[name()='svg']")))
    desktop.click()
    
    driver.find_element(By.XPATH, "//li[3]//span[1]//button[1]//*[name()='svg']").click()
    driver.find_element(By.XPATH, "//label[@for='tree-node-notes']//span[@class='rct-checkbox']//*[name()='svg']").click()
    driver.find_element(By.XPATH, "//label[@for='tree-node-wordFile']//span[@class='rct-checkbox']//*[name()='svg']").click()

    print("Toggle with different options clicked")
    time.sleep(t)
    
except Exception as ex: 
    print("ERROR:", type(ex).__name__)
    print("Message:", ex)


driver.find_element(By.XPATH, "//span[normalize-space()='Web Tables']").click()


try:
    rows = Select(WebDriverWait(driver,5).until(EC.element_to_be_clickable((By.XPATH, "//select[@aria-label='rows per page']"))))
    #driver.execute_script("arguments[0].scrollIntoView()",rows)
    rows.select_by_index(0) #5 rows
    
    delete = driver.find_element(By.XPATH, "//input[@id='searchBox']").send_keys("Cierra" + Keys.ENTER)

    print("5 ROWS")
    time.sleep(t)
    
except Exception as ex: 
    print("ERROR:", type(ex).__name__)
    print("Message:", ex)
    
driver.close()
```
