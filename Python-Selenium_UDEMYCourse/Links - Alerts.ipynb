```python
from selenium import webdriver 
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.service import Service
from selenium.webdriver.common.by import By

from selenium.webdriver.chrome.service import Service
from selenium.webdriver.firefox.service import Service
from selenium.webdriver.firefox.options import Options


from selenium.webdriver.support.ui import WebDriverWait, Select
from selenium.webdriver.support import expected_conditions as EC

from webdriver_manager.chrome import ChromeDriverManager
import time
```

```python
######################### Links ##############################
t = 2
driver = webdriver.Chrome()
driver.get("https://demoqa.com/text-box")
driver.maximize_window()
time.sleep(t)

driver.execute_script("window.scrollTo(0, 300)")
driver.find_element(By.XPATH, "//div[@class='element-list collapse show']//li[@id='item-5']").click()

links = driver.find_elements(By.TAG_NAME, "a")

print(f'Links: {len(links)}')

for num in links:
    print(num.text)
    
driver.execute_script("window.scrollTo(0, 500)")

driver.find_element(By.LINK_TEXT, "Created").click()
driver.find_element(By.LINK_TEXT, "Moved").click()
driver.find_element(By.LINK_TEXT, "Unauthorized").click()
driver.find_element(By.LINK_TEXT, "Not Found").click()

time.sleep(t)
driver.close()
```

```python
######################### Alerts ##############################
t = 1
driver = webdriver.Chrome()
driver.get("https://demoqa.com/text-box")
driver.maximize_window()
time.sleep(t)

driver.find_element(By.XPATH, "//body/div[@id='app']/div[contains(@class,'body-height')]/div[contains(@class,'container playgound-body')]/div[contains(@class,'row')]/div[contains(@class,'col-md-3')]/div[contains(@class,'left-pannel')]/div[contains(@class,'accordion')]/div[1]/span[1]/div[1]/div[2]").click()

try:
    alerts = WebDriverWait(driver, 5).until(EC.element_to_be_clickable((By.XPATH, "//*[@id='app']/div/div/div/div[1]/div/div/div[3]/span/div/div[2]")))
    driver.execute_script("arguments[0].scrollIntoView(true);", alerts)
    driver.execute_script("arguments[0].click();", alerts)

    alerts2 = WebDriverWait(driver, 5).until(EC.element_to_be_clickable((By.XPATH, "//span[normalize-space()='Alerts']")))
    alerts2.click()
    
    driver.execute_script("window.scrollBy(0, 100)")
     
    driver.find_element(By.XPATH, "//button[@id='alertButton']").click()
    time.sleep(t)
    # Wait for alert
    WebDriverWait(driver, 5).until(EC.alert_is_present())
    alert1 = driver.switch_to.alert
    time.sleep(t)
    alert1.accept()
    
    
    driver.find_element(By.XPATH, "//button[@id='timerAlertButton']").click()
    time.sleep(t)
    # Wait for alert
    WebDriverWait(driver, 10).until(EC.alert_is_present())
    alert2 = driver.switch_to.alert
    time.sleep(t)
    alert2.accept() #accept alert
    
    
    driver.find_element(By.XPATH, "//button[@id='confirmButton']").click()
    time.sleep(t)
    # Wait for alert
    WebDriverWait(driver, 5).until(EC.alert_is_present())
    alert3 = driver.switch_to.alert
    time.sleep(t)
    alert3.dismiss() #Cancel alert
    
    
    driver.find_element(By.XPATH, "//button[@id='promtButton']").click()
    time.sleep(t)
    # Wait for alert
    WebDriverWait(driver, 5).until(EC.alert_is_present())
    alert4 = driver.switch_to.alert   
    time.sleep(t)
    alert4.send_keys("Pablo Pombo")
    time.sleep(t)
    alert4.accept()


except Exception as ex:   #takes any kind of error 
    print("ERROR:", type(ex).__name__)
    print("Message:", ex)

    
time.sleep(t)
driver.close()
```
