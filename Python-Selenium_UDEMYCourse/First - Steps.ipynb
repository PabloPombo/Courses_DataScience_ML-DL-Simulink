```python
from selenium import webdriver 
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.service import Service
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.firefox.service import Service
from selenium.webdriver.firefox.options import Options


from webdriver_manager.chrome import ChromeDriverManager
```

```python
#pip install webdriver-manager
```

```python
######################### With Chrome ##############################

options = webdriver.ChromeOptions()

# Solo instala automáticamente la versión compatible de ChromeDriver
driver = webdriver.Chrome(
    service=Service(ChromeDriverManager().install()),
    options=options
)

driver.get("https://www.google.com")
print(driver.title)
driver.quit()
```

```python
######################### With Chrome ##############################

driver = webdriver.Chrome()

driver.get("https://aulavirtual.aevirtualclass.com/login/index.php")

print("Bienvenido a Selenium")
print(driver.title)

driver.quit()
```

```python
######################### With firefox ##############################
driver = webdriver.Firefox()

driver.get("https://aulavirtual.aevirtualclass.com/login/index.php")

print("Bienvenido a Selenium")
print(driver.title)

driver.close()
```

```python
######################### With firefox ##############################

# Ruta al geckodriver
service = Service(r"C:\Drivers\geckodriver32.exe")

# Configuración de Firefox
options = Options()
options.binary_location = r"C:\Program Files\Mozilla Firefox\firefox.exe"

# Inicializar driver
driver = webdriver.Firefox(service=service, options=options)

driver.get("https://www.google.com")
print(driver.title)
driver.quit()
```
